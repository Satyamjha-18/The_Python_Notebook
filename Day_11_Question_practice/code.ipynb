### Problem 1 : Student Grade

Problem Statement: Write a Python program that takes marks of 5 subjects as input.
Your program should:

Store the marks in a list.
Calculate the average marks.
Find the highest and lowest marks.
Display the grade using:
| Average | Grade |
| -------- | ----- |
| 90+ | A |
| 75-89 | B |
| 60-74 | C |
| Below 60 | D |


```python
marks = [95,70,80,85,90]

avg = sum(marks)/len(marks)
print(avg)
if avg >=90:
    print("Grade :","A")
elif avg >75:
    print("Grade :","B") 

elif avg >=60 :
    print("Grade :","C")

else :
    print("Grade :","C")

print(min(marks))
print(max(marks))
```

#### Word Frequency Counter

Problem Statement: Take a sentence as input.
Display how many times each word appears.
* Eg1: python is fun python is easy
* op: python - 2, is - 2 , easy - 1,

```python
s = input("Enter your words")

words= s.split(" ")
freq = {}

for word in words:
    if word not in freq:
        freq[word]=1

    else:
        freq[word]+=1
    
print(s)
print(words)
print(freq)
```

#### Character Counter 

Problem Statement: Take a string as input from the user.

 Display:
* 1.Number of vowels
* 2.Number of consonants
* 3.Number of digits
* 4.Number of spaces

```python
s = input()
s1=s.lower()  # esliye kiye taki search capital latter me na krna pade

vowels=0
consonants=0
digits=0
spaces=0
for c in s1 :
    if c in "aeiou":
        vowels+=1
    elif c>="a" and c <="z":
        consonants+=1
    elif c>="0"and c <="9":
        digits+=1

    elif c== " ":
        spaces+=1

print("vowels :",vowels)
print("consonants:",consonants)
print("digits :",digits)
print("spaces",spaces) 

```

### Password Validator

Problem Statement:A password is valid if it contains
minimum 8 characters
 * at least one uppercase letter
 * at least one lowercase letter
 * at least one digit

Display
* Strong Password or Weak Password


```python
passoword = "Sanya@1123"

upper=False
lower=False
digit=False

n =len(passoword)
for i in passoword:
    if i.islower():
        lower=True
    elif i.isupper():
        upper=True
    elif i.isdigit():
        digit=True

if n >=8  and upper ==True and lower== True and digit== True:
    print("Strong Password")
else:
    print("Wrong Password")



```
