### Remove key-value pair
* 1.pop
* 2.popitem
* 3.del
* clear

```python
# pop

d={'name': 'nitish', 'age': 25, 3: 4, 'gender': 'Male', 'weight': 70}
d.pop(3)
print(d)
```

```python
d.popitem() # delete last key value pair
print(d)
```

```python
# del 
d={'name': 'nitish', 'age': 25, 'gender': 'Male', 'weight': 70}
print(d)

del d['name']
print(d)
```

```python
# clear 
d.clear()
print(d)
```
