```python
numbers = [1,2,2,3,4,4,5]

unique=set(numbers)

even =set()

for num in unique:
    if num%2==0:
        even.add(num)

print(len(even))
```

##### Symmetric Difference

```python
A = {1, 2, 3}
B = {2, 3, 4}

print(A.symmetric_difference(B))
```

#### issubset()

```python
A = {1, 2}
B = {1, 2, 3, 4}

print(A.issubset(B))
```

#### issuperset()

```python
A = {1, 2, 3, 4}
B = {1, 2}

print(A.issuperset(B))
```

#### isdisjoint()

```python
A = {1, 2}
B = {3, 4}

print(A.isdisjoint(B))
```

#### copy()

```python
A = {10, 20, 30}

B = A.copy()

print(B)
```

#### Frozen Set

```python
A = frozenset([1, 2, 3, 4])

print(A)
```

#### Set Comprehension

```python
square = {x*x for x in range(1, 6)}

print(square)
```

#### Remove Duplicates from List

```python
numbers = [10, 20, 30, 20, 10, 40]

unique = list(set(numbers))

print(unique)
```

#### Common Elements

```python
list1 = [10, 20, 30, 40]
list2 = [30, 40, 50, 60]

common = set(list1).intersection(set(list2))

print(common)
```

#### Unique Elements

```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A.symmetric_difference(B))
```

```python
python = {"Amit", "Rahul", "Satyam", "Rohit"}

java = {"Rahul", "Rohit", "Ankit"}

print("Python Students:", python)
print("Java Students:", java)

print("Common Students:", python.intersection(java))

print("Only Python:", python.difference(java))

print("All Students:", python.union(java))
```
