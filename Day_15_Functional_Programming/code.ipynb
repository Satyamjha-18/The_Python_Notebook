### Functional programming (lambda function):

```python
n=[1,2,3,4,5]
```

```python
# Create a list to store a squre of each of the  number in the list

sq=[]

for elem in n:
    sq.append(elem**2)

print(sq)
```

```python
# by using functional programming( without writing a loop)

sq=list(map(lambda x: x**2,n))
sq
```

### Lambda function:Anonymous function

```python
def sq(x):
    return x*x
print(sq(5))
```

```python
sq=lambda x: x*x
sq(5)
```

### Map:
* map(function,iterable)--> it return a map object --> convert in into list --> list(map)

```python
# write a fun to return a list of the cube of evey num in the given list
def cube(lst):
    cb=[]
    for n in lst:
        cb.append(n**3)

    return cb
```

```python
cube([2,3,4])
```

```python
def cb(n):
    return n**3

cube=[]

lst =[4,6,2]
for elem in lst:
    cube.append(cb(elem))

print(cube)
```

```python
lst =[4,6,2]
cube =list(map(lambda x: x**3,lst))
print(cube)
```

### Filter 

```python
num =[1,2,3,4,5,6]

# write a code a list of all the even number of the given list

even =[]

for i in num :
    if i % 2==0:
        even.append(i)

print(even)

```

```python
num =[1,2,3,4,5,6]

even = list(filter(lambda x: x%2==0,num))
print(even)
```

```python
odd = list(filter(lambda x: x%2!=0,[1,2,3,4,5,6]))
print(odd)
```
