```python
# write a function to print Grade of Every student 

def Cal_grade(marks):
    if marks >90:
      grade='A'
    elif marks >75:
      grade='B'
    elif marks>60:
       grade='C'
    else:
       grade='D'

    return grade      
       
         
```
